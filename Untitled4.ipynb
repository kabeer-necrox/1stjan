```python



students = []
def create_student(name):
    return {'name': name, 'grades': {}}
```

```python
def add_new_student(name):
    # Create a new student template
    student_dict = create_student(name)
    # Append the student dictionary to the students list
    students.append(student_dict)

```

```python
def assign_grade(student_name, subject, grade):
    for student in students:
        if student['name'] == student_name:
            student['grades'][subject] = grade
            break
```

```python
def calculate_student_average(student_name):
    # Find the student in the list
    target_student = None
    for student in students:
        if student['name'] == student_name:
            target_student = student
            break

    # If student is not found, print an error
    if target_student is None:
        print(f"Error: Student '{student_name}' not found.")
        return None

    grades = target_student['grades']

    # Handle the case where the student has no grades
    if not grades:
        print(f"{student_name} has no grades assigned yet.")
        return 0

    # Calculate and return average
    total = sum(grades.values())
    average = total / len(grades)
    return average
```

```python
def display_student_info(student_name):
    # Find the student in the list
    target_student = None
    for student in students:
        if student['name'] == student_name:
            target_student = student
            break

    # If student is not found, print an error
    if target_student is None:
        print(f"Error: Student '{student_name}' not found.")
        return

    # Display student name and all subject grades
    print(f"Student Name: {target_student['name']}")
    grades = target_student['grades']
    if not grades:
        print("No grades assigned yet.")
    else:
        for subject, grade in grades.items():
            print(f"Subject: {subject}, Grade: {grade}")
```

```python
def main_menu():
    while True:
        print("\n--- Main Menu ---")
        print("1. Add New Student")
        print("2. Exit")

        choice = input("Enter your choice (1 or 2): ")

        if choice == '1':
            name = input("Enter student name: ")
            add_new_student(name)
            print(f"Student '{name}' added successfully!")
        elif choice == '2':
            print("Exiting the program. Goodbye!")
            break
        else:
            print("Invalid choice. Please choose 1 or 2.")
```
