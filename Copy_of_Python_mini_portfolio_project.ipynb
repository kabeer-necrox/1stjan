# Python Mini Portfolio Project: Student Grade Management System

| Task 2   | Add Students to the System                 | 5     |
| Task 3   | Assign a Single Grade to a Student         | 5     |
| Task 4   | Calculate Student's Average Grade          | 5     |

| Task 6   | Basic Menu Loop for Adding Students        | 5     |

Marks will be awarded based on correctness, adherence to requirements, and basic code readability.

### Task 1: Initialize Data Structures and Create Student Template

**Objective**: Set up the basic storage for student information.

*   **Concepts**: Variables, Lists, Dictionaries.
*   **Requirements**:
    *   Initialize an empty list called `students`.
    *   Define a function `create_student(name)` that returns a dictionary for a new student, like `{'name': name, 'grades': {}}`.

### Task 2: Add Students to the System

**Objective**: Implement a simple way to add new students to our `students` list.

*   **Concepts**: Functions, List `append()`.
*   **Requirements**:
    *   Create a function `add_new_student(name)` that takes a student's name.
    *   Inside the function, use `create_student(name)` (from Task 1) and append the new student dictionary to the `students` list.
    *   No duplicate checking is needed for this task.

### Task 3: Assign a Single Grade to a Student

**Objective**: Write a function to add a grade for a student in a specific subject.

*   **Concepts**: Functions, Dictionaries (accessing and modifying nested dictionaries).
*   **Requirements**:
    *   Create a function `assign_grade(student_name, subject, grade)`.
    *   Find the student by `student_name` in the `students` list.
    *   If the student is found, add the `grade` for the `subject` to their `grades` dictionary.
    *   Assume the `grade` input is valid for now (e.g., a number).

### Task 4: Calculate Student's Average Grade

**Objective**: Write a function to compute the average grade for a specific student.

*   **Concepts**: Functions, Loops (`for`), Arithmetic Operators, Conditional Statements (handling no grades).
*   **Requirements**:
    *   Create a function `calculate_student_average(student_name)`.
    *   Find the student. If not found, print an error.
    *   If found, calculate the average of all grades in their `grades` dictionary.
    *   Handle the case where a student has no grades.

### Task 5: Display a Student's Basic Information

**Objective**: Create a function to show a student's name and all their grades.

*   **Concepts**: Functions, Loops (`for`), Conditional Statements, String Formatting.
*   **Requirements**:
    *   Create a function `display_student_info(student_name)`.
    *   Find the student. If not found, print an error.
    *   If found, print their name and then list each subject and its grade.
    *   Do not include the average or pass/fail status in this task.

### Task 6: Basic Menu Loop for Adding Students

**Objective**: Create a simple interactive menu where users can add students and exit the program.

*   **Concepts**: `While` Loop, Conditional Statements, User Input (`input()`), Calling Functions.
*   **Requirements**:
    *   Create a `main_menu()` function that displays options: "1. Add New Student", "2. Exit".
    *   Use a `while` loop to keep the menu running until the user chooses to exit.
    *   If "Add New Student" is chosen, prompt for a name and call the `add_new_student()` function from Task 2.
    *   Print a message for invalid menu choices.
