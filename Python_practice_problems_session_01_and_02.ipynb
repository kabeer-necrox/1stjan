Assignment operators. A water tank holds 500 litres. In one day 120 litres are used, then 80 litres are refilled, then the level is doubled by connecting a second tank. Using only augmented operators (-=, +=, *=) on a variable litres, print the final amount.


```python


litres = 500

litres -= 120
litres += 80
litres *= 2
```
